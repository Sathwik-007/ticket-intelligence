```python
import numpy as np
import pickle as pkl
import time
import faiss
```

# Load Artifacts

```python
X_embed = np.load("../data/processed/embeddings.npy")
```

# Use HDBSCAN wrapper

```python
from vector_store import TicketIndexer

print("Building HNSW Graph Index...")
indexer = TicketIndexer(vector_dim=X_embed.shape[1], use_hnsw=True)
indexer.build_index(X_embed)
print(f"Total vectors indexed: {indexer.index.ntotal}")
```

```python
mock_query = X_embed[21:22]
```

```python
start_time = time.perf_counter()
distances, indices = indexer.search(mock_query, k=5)
end_time = time.perf_counter()

print(f"\nRetrieval Time: {(end_time - start_time) * 1000:.4f} milliseconds")
print(f"Top 5 Indices: {indices}")
print(f"L2 Distances: {distances}")
```

# Save Faiss Index to disk

```python
faiss.write_index(indexer.index, "../data/processed/faiss_full.index")
print("Faiss index write to disk success ✅")
```
